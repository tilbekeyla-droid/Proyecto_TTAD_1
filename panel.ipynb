```python
import json

with open("data.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

ventas = datos["ventas"]
```

```python
ingresos = {}

for venta in ventas:
    producto = venta["producto"]
    importe = venta["cantidad"] * venta["precio_unitario"]

    if producto not in ingresos:
        ingresos[producto] = 0

    ingresos[producto] += importe

producto_top = max(ingresos, key=ingresos.get)

print("Producto con más ingresos:", producto_top)
print("Ingresos:", ingresos[producto_top])
```

```python
vendedor = "Ana"

for venta in ventas:
    if venta["vendedor"] == vendedor:
        print(venta)
```

```python
total = 0

for venta in ventas:
    total += venta["cantidad"] * venta["precio_unitario"]

ticket_medio = total / len(ventas)

print("Ticket medio:", ticket_medio)
```
